```python
def test_query():
    api = overpy.Overpass(url="https://maps.mail.ru/osm/tools/overpass/api/", retry_timeout=120)

    # fetch all ways and nodes
    result = api.query("""node["name"="Gielgen"];out body;""")

    for way in result.ways:
        print(way)

test_query()
```
