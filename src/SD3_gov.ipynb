```python
gov_gdf = gpd.GeoDataFrame([a.dict for a in GOV_AGENCIES], crs=4326).to_crs(3414)
```

```python
def export(gov_gdf, fname: str = "gov_workforce.parquet"):
    fpath_intermediate = data_filepaths.path_data_intermediate
    fpath_gov = os.path.join(fpath_intermediate, fname)

    gov_gdf = gov_gdf.copy()

    gov_gdf = gov_gdf.rename(columns={"workforce": "workers"})
    gov_gdf = gov_gdf[["name", "workers", "geometry"]]

    gov_gdf.to_parquet(fpath_gov)
    return gov_gdf


export(gov_gdf, "gov_workforce_v1.parquet").head()
```
