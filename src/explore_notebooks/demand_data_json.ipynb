# Schema

https://www.subwaybuilder.com/docs/api-reference/cities

## demand_data.json

### `points`

`id`

* (STRING)
* can just be an integer (e.g. 150030086343000)
* or special ids:
  * `AIR_<terminal-code>`
  * `UNI_<index>` - in vanilla maps this just a number
  * In Depot the full list seems to be: `AIR`,`AMU`,`AQU`,`BTH`,`CNV`,`CUL`,`EVT`,`EXT`,`GOV`,`HER`,`HOS`,`LIB`,
    `MIL`,`MUS`,`NAT`,`PORT`,`PRK`, `REL`,`RST`,`SCH`,`SHP`,`SPO`,`UNI`,`ZOO`,
    `nature_park`,`lake`,`arena`,`racetrack`,`sports_complex`, `sports_park`,`stadium`,

`location`: [lon, lat]

`jobs`: int

`residents`: int

`popIds`: list[str] of all associated pops with this point, whether they be residence or job

### `pops`

`id`: STRING, can just be integers

`size`: on vanilla maps, these are all `200`

`residenceId`: FK to `points.id`

`jobId`: FK to `points.id`

`drivingSeconds`

`drivingDistance`
